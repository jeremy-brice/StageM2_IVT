# MA Domain & Period 

```python
u = xr.open_dataset('/bettik/PROJECTS/pr-regional-climate/santolam/era5/monthly/Ulevs_1000-001_mon.era5_194801_202305_interp.nc')['u']
v = xr.open_dataset('/bettik/PROJECTS/pr-regional-climate/santolam/era5/monthly/Vlevs_1000-001_mon.era5_194801_202305_interp.nc')['v']
```

```python
v.attrs = {}  # les atributs pas utiles pour les calculs
#v = v.load()  # charger en mémoire pour pas avoir à relire le fichier pour les calculs
```

```python
iyear=1960
fyear=2022
```

```python
u = u.sel(time=slice(f"{iyear}-01-01", f"{fyear}-12-31")).sel(level=500)
v = v.sel(time=slice(f"{iyear}-01-01", f"{fyear}-12-31")).sel(level=500)
```

```python
#focus on Middle Asia (MA)
u_MA = field_dom(u,'NHe')
v_MA = field_dom(v,'NHe')
```

```python
wind = np.sqrt(u_MA**2 + v_MA**2)
```

# wind Mean 

```python
u_mean = u_MA.mean(dim='time', skipna=True)
v_mean = v_MA.mean(dim='time', skipna=True)
wind_mean = wind.mean(dim='time', skipna=True)
```

```python
fig, ax = plt.subplots(subplot_kw={'projection': ccrs.PlateCarree()}, figsize=(12,6))

#tp color definition
vmax = 50
#vmax = tp_mean.max()
levels = np.arange(0, vmax + 5, 5)
cmap = plt.get_cmap("viridis_r", len(levels)-1)
norm = mcolors.BoundaryNorm(levels, cmap.N)
pcm = ax.pcolormesh(
    u_mean['lon'], u_mean['lat'], wind_mean,
    cmap=cmap, norm=norm,transform=ccrs.PlateCarree(), alpha=0.7)

#wind Vectors
step = 3
u_plot = u_mean[::step, ::step]
v_plot = v_mean[::step, ::step]
q = ax.quiver(u_plot['lon'],u_plot['lat'],u_plot,v_plot,transform=ccrs.PlateCarree(),scale=800,width=0.0025,headwidth=3
             ,rasterized=True)

#Borders
ax.coastlines(linewidth=1.5)
ax.add_feature(cfeature.BORDERS, linewidth=1.5)

lon_min = float(u_mean['lon'].min())
lon_max = float(u_mean['lon'].max())
lat_min = float(u_mean['lat'].min())
lat_max = float(u_mean['lat'].max())
ax.set_extent([lon_min, lon_max, lat_min, lat_max], crs=ccrs.PlateCarree())

#Grid of lat & lon
gl = ax.gridlines(draw_labels=True, linewidth=1, color='gray', alpha=0.5, linestyle='--')
gl.top_labels = False
gl.right_labels = False
gl.xlabel_style = {"size": 12}
gl.ylabel_style = {"size": 12}

#Colorbar & title
cbar_ax = fig.add_axes([0.25, 0.01, 0.5, 0.03])
cbar = fig.colorbar(pcm, cax=cbar_ax, orientation='horizontal', extend='max')
cbar.set_label("u (m/s)", fontsize=14)
cbar.ax.tick_params(labelsize=12)
ax.quiverkey(q,0.95, -0.08,    50,"wind\n50 m s⁻¹",labelpos='S',coordinates='axes')
ax.set_title(f"Wind mean over NHe ({iyear}-{fyear})", fontsize=16, weight='bold')

#MA zone
latS, latN, lonW, lonE = coord_domain('MA')
rect = mpatches.Rectangle((lonW, latS), lonE - lonW, latN - latS,linewidth=2, edgecolor='red', facecolor='none',
                          alpha=0.7, transform=ccrs.PlateCarree())
ax.add_patch(rect)
#ax.text((lonW + lonE) / 2, latN + 3, "HMA",horizontalalignment='center',verticalalignment='center',fontsize=16,fontweight='bold',
#        color='black',transform=ccrs.PlateCarree(),zorder=7,bbox=dict(facecolor='white', alpha=0.8, edgecolor='black', pad=3))

#Save
#plotsDir="/home/bricej/figures/climbas_tp_era5/"
#plotname=f'tp_mean_MA_{iyear}-{fyear}'
#plt.savefig(f"{plotsDir}{plotname}.png", bbox_inches="tight")
#print(f"Saved: {plotsDir}{plotname}.png")

plt.show()
```

# Monthly Climatology of U

```python
#Group by month + mean
monthly_clim_u = u_MA.groupby('time.month').mean('time')
monthly_clim_v = v_MA.groupby('time.month').mean('time')
monthly_clim_wind = wind.groupby('time.month').mean('time')
```

```python
#vmax = float(monthly_clim.max())
levels = np.arange(0, 70, 5)
cmap = plt.get_cmap("viridis_r")
norm = mcolors.BoundaryNorm(levels, cmap.N)

fig, axes = plt.subplots(4, 3,figsize=(20, 14),subplot_kw={'projection': ccrs.PlateCarree()})
axes = axes.flatten()

for i, ax in enumerate(axes):
    data_month = monthly_clim_wind.sel(month=i+1)
    im = ax.pcolormesh(data_month.lon,data_month.lat,data_month,transform=ccrs.PlateCarree(),cmap=cmap,norm=norm,shading="auto", alpha=0.7)

    ax.add_feature(cfeature.COASTLINE, linewidth=1)
    ax.add_feature(cfeature.BORDERS, linestyle="-")

    #wind Vectors
    step = 3
    u_plot = monthly_clim_u.sel(month=i+1)[::step, ::step]
    v_plot = monthly_clim_v.sel(month=i+1)[::step, ::step]
    q = ax.quiver(u_plot['lon'],u_plot['lat'],u_plot,v_plot,transform=ccrs.PlateCarree(),scale=1000,width=0.0025,headwidth=3
                 ,rasterized=True)
    
    lon_min = float(data_month['lon'].min())
    lon_max = float(data_month['lon'].max())
    lat_min = float(data_month['lat'].min())
    lat_max = float(data_month['lat'].max())
    ax.set_extent([lon_min, lon_max, lat_min, lat_max], crs=ccrs.PlateCarree())
    
    ax.set_title(calendar.month_name[i+1],fontsize=14,weight="bold")

    gl = ax.gridlines(crs=ccrs.PlateCarree(),draw_labels=True,linewidth=0.8,color="gray",alpha=0.4,linestyle="--")
    gl.top_labels = False
    gl.right_labels = False
    gl.left_labels = (i % 3 == 0)
    gl.bottom_labels = (i >= 9)
    gl.xlabel_style = {"size": 12}
    gl.ylabel_style = {"size": 12}

cbar_ax = fig.add_axes([0.25, 0.05, 0.5, 0.025])
cbar = fig.colorbar(im, cax=cbar_ax, orientation="horizontal", extend='max')
cbar.set_label("u (m/s)", fontsize=14)
cbar.ax.tick_params(labelsize=12)
ax.quiverkey(q,0.70, -0.3,    70,"wind\n70 m s⁻¹",labelpos='S',coordinates='axes', fontproperties={'size': 13})

fig.subplots_adjust(left=0.08,right=0.92,top=0.93,bottom=0.12,wspace=0.05,hspace=0.18)
plt.suptitle(f"Monthly U Climatology over MA ({iyear}-{fyear})",fontsize=18,weight="bold")

#Save
#plotsDir="/home/bricej/figures/climbas_tp_era5/"
#plotname=f'monthly_tp_clim_MA_{iyear}-{fyear}'
#plt.savefig(f"{plotsDir}{plotname}.png", bbox_inches="tight")
#print(f"Saved: {plotsDir}{plotname}.png")

plt.show()
```

# Seasonal Climatology of U

```python
seasons = ["DJFM", "AM", "JJAS", "ON"]
start_year = iyear
end_year = fyear
clim_dict = {}
clim_dict_u = {}
clim_dict_v = {}


#Climatology for each season
for season in seasons:
    clim_season = clim(wind, season=season, imon=1, iyr=start_year, fmon=12, fyr=end_year)
    clim_dict[season] = clim_season
    clim_season_u = clim(u_MA, season=season, imon=1, iyr=start_year, fmon=12, fyr=end_year)
    clim_dict_u[season] = clim_season_u
    clim_season_v = clim(v_MA, season=season, imon=1, iyr=start_year, fmon=12, fyr=end_year)
    clim_dict_v[season] = clim_season_v
```

```python
seasons = ["DJFM", "AM", "JJAS", "ON"]
vmin = 0
vmax = 70
#vmax = max(ds["tp"].max().item() for ds in clim_dict.values())

levels = np.arange(vmin, vmax + 5, 5)
cmap = plt.get_cmap("viridis_r")
norm = mcolors.BoundaryNorm(levels, cmap.N)

fig, axes = plt.subplots(2, 2, figsize=(16, 10), subplot_kw={'projection': ccrs.PlateCarree()})
axes = axes.flatten()

for i, ax in enumerate(axes):
    season = seasons[i]
    data_season = clim_dict[season]

    im = ax.pcolormesh(data_season.lon,data_season.lat,data_season,transform=ccrs.PlateCarree(),cmap=cmap,norm=norm,shading="auto",
                      alpha=0.7)

    #wind Vectors
    step = 3
    u_plot = clim_dict_u[season][::step, ::step]
    v_plot = clim_dict_v[season][::step, ::step]
    q = ax.quiver(u_plot['lon'],u_plot['lat'],u_plot,v_plot,transform=ccrs.PlateCarree(),scale=1000,width=0.0025,headwidth=3
                 ,rasterized=True)
    
    ax.add_feature(cfeature.COASTLINE, linewidth=1)
    ax.add_feature(cfeature.BORDERS, linestyle="-")

    lon_min = float(data_season['lon'].min())
    lon_max = float(data_season['lon'].max())
    lat_min = float(data_season['lat'].min())
    lat_max = float(data_season['lat'].max())
    ax.set_extent([lon_min, lon_max, lat_min, lat_max], crs=ccrs.PlateCarree())
    
    ax.set_title(season, fontsize=16, weight="bold")

    gl = ax.gridlines(crs=ccrs.PlateCarree(),draw_labels=True,linewidth=0.8,color="gray",alpha=0.4,linestyle="--")
    gl.top_labels = False
    gl.right_labels = False
    gl.left_labels = (i % 2 == 0)
    gl.bottom_labels = (i >= 2) 
    gl.xlabel_style = {"size": 12}
    gl.ylabel_style = {"size": 12}

    #MA zone
    latS, latN, lonW, lonE = coord_domain('MA')
    rect = mpatches.Rectangle((lonW, latS), lonE - lonW, latN - latS,linewidth=2, edgecolor='red', facecolor='none',
                              alpha=0.7, transform=ccrs.PlateCarree())
    ax.add_patch(rect)

cbar_ax = fig.add_axes([0.1, 0.12, 0.7, 0.015])
cbar = fig.colorbar(im, cax=cbar_ax, orientation="horizontal", extend='max')
cbar.set_label("wind (m/s)", fontsize=14)
cbar.ax.tick_params(labelsize=12)
ax.quiverkey(q,0.85, -0.15,    70,"70 m s⁻¹",labelpos='S',coordinates='axes', fontproperties={'size': 14})

fig.subplots_adjust(left=0.08, right=0.92, top=0.93, bottom=0.12, wspace=0.1, hspace=-0.2)
plt.suptitle(f"Seasonal 500 mb wind Climatology over MA ({iyear}-{fyear})", fontsize=18, weight="bold")

#Save
#plotsDir="/home/bricej/figures/climbas_tp_era5/"
#plotname=f'seasonal_tp_clim_MA_{iyear}-{fyear}'
#plt.savefig(f"{plotsDir}{plotname}.png", bbox_inches="tight")
#print(f"Saved: {plotsDir}{plotname}.png")

plt.show()
```
