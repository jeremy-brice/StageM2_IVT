```python
import sys
sys.path.insert(1, '/home/bricej/MyPythonLibrary/StageM2_IVT/library/')
import domain
from domain import *
from climbas import *
import numpy as np
import xarray as xr
from netCDF4 import Dataset
import dask
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import matplotlib.patches as mpatches
from matplotlib.patches import Rectangle
import calendar
```

# Domain & Period

```python
iyear=1960
fyear=2024
nx=10;ny=10
dom='GRq'
dom_NST = 'GRc'
year_NST = '2000'
```

# Elevation data

```python
ds_grM= xr.open_dataset(f'/home/santolam/NST.{year_NST}.01.01.00.{dom_NST}.nc')
LONS=ds_grM.LON[ny:-ny,nx:-nx].values
LATS=ds_grM.LAT[ny:-ny,nx:-nx].values
dsSH=ds_grM.SH[ny:-ny,nx:-nx]
latS=LATS.min(); latN=LATS.max()
lonW=LONS.min(); lonE=LONS.max()

elev = dsSH
```

# Seasonal data

```python
SMB = xr.open_mfdataset(f'/bettik/PROJECTS/pr-regional-climate/santolam/MARout_post/{dom}/spin2/work/monthly/SMB_monsum_MARv3.14_ER5_spin2_{dom}_*.nc',
                       combine="by_coords")['SMB']

#Domain & period def
SMB = SMB.rename({'TIME': 'time'})
SMB = SMB[:,:,ny:-ny,nx:-nx].sel(time=slice(str(iyear),str(fyear)))

# .isel(SECTOR=0) = ICE domain / .isel(SECTOR=1) = partout ou presque je sais plus
SMB = SMB.isel(SECTOR=0)
```

```python
seasons = ["DJFM","AM","JJAS","ON"]

tp_dict = {}

for season in seasons:
    tp_season = clim(SMB, season=season, imon=1, iyr=iyear, fmon=12, fyr=fyear)
    tp_dict[season] = tp_season

tp_dict['DJFM'] = tp_dict['DJFM'] 
tp_dict['AM'] = tp_dict['AM'] 
tp_dict['JJAS'] = tp_dict['JJAS']
tp_dict['ON'] = tp_dict['ON']
```

```python
#==================== Color grading ============================

step = 25
vmax = 300
print(vmax)
levels = np.arange(-vmax, vmax + step, step)
base_cmap = plt.get_cmap("RdBu", len(levels)-1)
colors = base_cmap(np.arange(base_cmap.N))

# mettre blanc au milieu
# trouver l'indice du zéro
mid = base_cmap.N // 2
# remplacer les deux couleurs autour de 0 par du blanc
colors[mid-1] = [1, 1, 1, 1]
colors[mid]   = [1, 1, 1, 1]

cmap = mcolors.ListedColormap(colors)
norm = mcolors.BoundaryNorm(levels, ncolors=len(colors))

#===================== Plot each season ==================

fig, axes = plt.subplots(
    2, 2,
    figsize=(16, 10),
    subplot_kw={'projection': ccrs.PlateCarree()}
)
axes = axes.flatten()

for i, ax in enumerate(axes):
    season = seasons[i]
    data_model = tp_dict[season]
       
    im = ax.pcolormesh(
           LONS, LATS, data_model,
           transform=ccrs.PlateCarree(),
           cmap=cmap,
           norm=norm,
           shading="auto"
       )
        
    ax.set_extent([lonW, lonE, latS, latN], crs=ccrs.PlateCarree())

    ax.add_feature(cfeature.COASTLINE, linewidth=1, alpha=0.8)
    ax.add_feature(cfeature.BORDERS, linestyle="-", alpha=0.8)

    SH = ax.contour(
        LONS,
        LATS,
        elev,
        levels=[1000,2000,3000,4000,5000,6000,7000],
        colors="k",
        linewidths=0.5,
        alpha=0.6,
        transform=ccrs.PlateCarree()
    )
    ax.clabel(SH, inline=True, fontsize=9, fmt="%d")
    
    ax.set_title(season, fontsize=16, weight="bold")

    gl = ax.gridlines(crs=ccrs.PlateCarree(),draw_labels=True,linewidth=0.8,color="gray",alpha=0.4,linestyle="--")
    gl.top_labels = False
    gl.right_labels = False
    gl.left_labels = (i % 2 == 0)
    gl.bottom_labels = (i >= 2) 
    gl.xlabel_style = {"size": 12}
    gl.ylabel_style = {"size": 12}

#==================== Bar color ============================

cbar_ax = fig.add_axes([0.1, 0.05, 0.8, 0.025])
cbar = fig.colorbar(im, cax=cbar_ax, orientation="horizontal", extend='both')
cbar.set_label("SMB (mm w.e. / month)", fontsize=20)
cbar.ax.tick_params(labelsize=20)

#==================== Final adjustments ============================

fig.subplots_adjust(left=0.08, right=0.92, top=0.96, bottom=0.05, wspace=0.05, hspace=-0.2)
plt.suptitle(
    f"Seasonal SMB over {dom} ({iyear}-{fyear})",
    fontsize=18,
    weight="bold"
)

plt.show()
```

```python
seasons = {"DJFM": [12,1,2,3], 
           "AM": [4,5], 
           "JJAS": [6,7,8,9], 
           "ON": [10,11]}

trend_dict = {}
p_dict={}

for season in seasons:
    tp_season = seasonal_selection2(SMB, seasons[season], iyear, fyear)
    tp_season = tp_season.compute()
    par = trend_vect(tp_season.time, tp_season, dim='time')
    trend=par[0]*10; p=par[3]
    trend_dict[season]=trend
    p_dict[season]=p
```

```python
seasons = ['DJFM','AM','JJAS','ON']

#==================== Color grading ============================

step = 5
vmax = 50
print(vmax)
levels = np.linspace(-vmax, vmax, int(2*vmax/step)+1)
base_cmap = plt.get_cmap("RdBu", len(levels)-1+2)
colors = base_cmap(np.arange(base_cmap.N))

# mettre blanc au milieu
# trouver l'indice du zéro
mid = base_cmap.N // 2
# remplacer les deux couleurs autour de 0 par du blanc
colors[mid-1] = [1, 1, 1, 1]
colors[mid]   = [1, 1, 1, 1]

cmap = mcolors.ListedColormap(colors)
norm = mcolors.BoundaryNorm(levels, ncolors=len(colors), extend='both')

#===================== Plot each season ==================

fig, axes = plt.subplots(
    2, 2,
    figsize=(16, 10),
    subplot_kw={'projection': ccrs.PlateCarree()}
)
axes = axes.flatten()

for i, ax in enumerate(axes):
    season = seasons[i]
    data_model = trend_dict[season]
    
    im = ax.pcolormesh(
           LONS, LATS, data_model,
           transform=ccrs.PlateCarree(),
           cmap=cmap,
           norm=norm,
           shading="auto"
       )

    sig = p_dict[season] < 0.05   
    step = 1
    mask = sig.values

    ax.scatter(
        LONS[mask],
        LATS[mask],
        color='black', s=2,
        transform=ccrs.PlateCarree(),
        zorder=5, alpha=0.4
    )
    
    ax.set_extent([lonW, lonE, latS, latN], crs=ccrs.PlateCarree())

    ax.add_feature(cfeature.COASTLINE, linewidth=1, alpha=0.8)
    ax.add_feature(cfeature.BORDERS, linestyle="-", alpha=0.8)

    SH = ax.contour(
        LONS,
        LATS,
        elev,
        levels=[1000,2000,3000,4000,5000,6000,7000],
        colors="k",
        linewidths=0.5,
        alpha=0.6,
        transform=ccrs.PlateCarree()
    )
    ax.clabel(SH, inline=True, fontsize=9, fmt="%d")
    
    ax.set_title(season, fontsize=16, weight="bold")

    gl = ax.gridlines(crs=ccrs.PlateCarree(),draw_labels=True,linewidth=0.8,color="gray",alpha=0.4,linestyle="--")
    gl.top_labels = False
    gl.right_labels = False
    gl.left_labels = (i % 2 == 0)
    gl.bottom_labels = (i >= 2) 
    gl.xlabel_style = {"size": 12}
    gl.ylabel_style = {"size": 12}

#==================== Bar color ============================

cbar_ax = fig.add_axes([0.1, 0.05, 0.8, 0.025])
cbar = fig.colorbar(im, cax=cbar_ax, orientation="horizontal", extend='both')
cbar.set_label("SMB (mm w.e. month⁻¹ / decade)", fontsize=20)
cbar.ax.tick_params(labelsize=20)

#==================== Final adjustments ============================

fig.subplots_adjust(left=0.08, right=0.92, top=0.96, bottom=0.05, wspace=0.05, hspace=-0.2)
plt.suptitle(
    f"Seasonal SMB trend over {dom} ({iyear}-{fyear})",
    fontsize=18,
    weight="bold"
)

plt.show()
```
