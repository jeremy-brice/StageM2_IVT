```python
ds_grM= xr.open_dataset(f'/home/santolam/NST.{year_NST}.01.01.00.{dom_NST}.nc')
LONS=ds_grM.LON[ny:-ny,nx:-nx].values
LATS=ds_grM.LAT[ny:-ny,nx:-nx].values
dsSH=ds_grM.SH[ny:-ny,nx:-nx]
latS=LATS.min(); latN=LATS.max()
lonW=LONS.min(); lonE=LONS.max()

elev = dsSH
elev = elev.rename({'x': 'X', 'y': 'Y'})
```

# SF & RF & TF Data

```python
RF = xr.open_mfdataset(f'/bettik/PROJECTS/pr-regional-climate/santolam/MARout_post/{dom}/spin2/work/monthly/RF_monsum_MARv3.14_ER5_spin2_{dom}_*.nc',
                       combine="by_coords")['RF']
RF = RF.rename({'TIME': 'time'})
RF = RF[:,ny:-ny,nx:-nx].sel(time=slice(f"{iyear}-01-01", f"{fyear}-12-31"))

SF = xr.open_mfdataset(f'/bettik/PROJECTS/pr-regional-climate/santolam/MARout_post/{dom}/spin2/work/monthly/SF_monsum_MARv3.14_ER5_spin2_{dom}_*.nc',
                       combine="by_coords")['SF']
SF = SF.rename({'TIME': 'time'})
SF = SF[:,ny:-ny,nx:-nx].sel(time=slice(f"{iyear}-01-01", f"{fyear}-12-31"))


#Compute Total Rainfall
TF = RF + SF
```

# Glaciers only

```python
dsICE = ds_grM.ICE[ny:-ny,nx:-nx]
dsICE = dsICE.rename({'x': 'X', 'y': 'Y'})
dsICE.plot()
```

```python
ICE_tot = dsICE.sum()
print(ICE_tot)
```

# Mean over period

```python
#Grille pas exactement les mêmes valeurs des fois à ,e-5 près 
dsICE = dsICE.assign_coords({
    "X": SF.X,
    "Y": SF.Y
})
SF_mean = SF.mean('time', skipna=True)
SF_mean = (SF_mean * dsICE)

dsICE = dsICE.assign_coords({
    "X": RF.X,
    "Y": RF.Y
})
RF_mean = RF.mean('time', skipna=True)
RF_mean = (RF_mean * dsICE)

dsICE = dsICE.assign_coords({
    "X": TF.X,
    "Y": TF.Y
})
TF_mean = TF.mean('time', skipna=True)
TF_mean = (TF_mean * dsICE)
```

# Mean per elevation bands (step = 250 m)

```python
# Altitude bands per 250 m
elev_bands = np.arange(2500, 6001, 250)

#Grille pas exactement les mêmes valeurs des fois à ,e-5 près 
elev = elev.assign_coords({
    "X": dsICE.X,
    "Y": dsICE.Y
})
# ICE tot per elev band
ICE_tot_bands = dsICE.groupby_bins(elev, bins=elev_bands).sum()
```

```python
#Grille pas exactement les mêmes valeurs des fois à ,e-5 près 
elev = elev.assign_coords({
    "X": RF.X,
    "Y": RF.Y
})
# RF mean per band of 250 m
RF_bands = RF_mean.groupby_bins(elev, bins=elev_bands).sum() / ICE_tot_bands
```

```python
#Grille pas exactement les mêmes valeurs des fois à ,e-5 près 
elev = elev.assign_coords({
    "X": SF.X,
    "Y": SF.Y
})

# RF mean per band of 250 m
SF_bands = SF_mean.groupby_bins(elev, bins=elev_bands).sum() / ICE_tot_bands
```

```python
#Grille pas exactement les mêmes valeurs des fois à ,e-5 près 
elev = elev.assign_coords({
    "X": TF.X,
    "Y": TF.Y
})

# RF mean per band of 250 m
TF_bands = TF_mean.groupby_bins(elev, bins=elev_bands).sum()
TF_bands = TF_bands / ICE_tot_bands
```

```python
R_bands = SF_bands / TF_bands * 100
```

```python
y = [interval.right for interval in RF_bands.SH_bins.values]

fig, ax = plt.subplots(figsize=(5,8))

ax.plot(
    RF_bands.values, y, 
    marker='o', 
    linestyle='-',
    color='royalblue',
    label=f'RF',
    linewidth=2)

ax.plot(
    SF_bands.values, y, 
    marker='o', 
    linestyle='-',
    color='violet',
    label=f'SF',
    linewidth=2)

ax.plot(
    TF_bands.values, y, 
    marker='o', 
    linestyle='--',
    color='green',
    label=f'TF',
    linewidth=2, alpha = 0.6)

ax.grid(True, linestyle='--', alpha=0.5, zorder=0)
ax.legend(fontsize=12)

ax.set_xlabel('Precipitation (mm / month)')
ax.set_ylabel('Elevation (m)')

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(5,8))

ax.plot(
    R_bands.values, y, 
    marker='o', 
    linestyle='-',
    color='red',
    label=f'R',
    linewidth=2)

ax.grid(True, linestyle='--', alpha=0.5, zorder=0)
ax.legend(fontsize=12)

ax.set_xlabel('R (%)')
ax.set_ylabel('Elevation (m)')

plt.tight_layout()
plt.show()
```

# Monthly mean vertical profile

```python
SF.isel(time=5).plot()
```

```python
dsICE = dsICE.assign_coords({
    "X": SF.X,
    "Y": SF.Y
})
SF_monthly_mean = SF.groupby('time.month').mean('time', skipna=True)
SF_monthly_mean = (SF_monthly_mean * dsICE)

dsICE = dsICE.assign_coords({
    "X": RF.X,
    "Y": RF.Y
})
RF_monthly_mean = RF.groupby('time.month').mean('time', skipna=True)
RF_monthly_mean = (RF_monthly_mean * dsICE)

dsICE = dsICE.assign_coords({
    "X": TF.X,
    "Y": TF.Y
})
TF_monthly_mean = TF.groupby('time.month').mean('time', skipna=True)
TF_monthly_mean = (TF_monthly_mean * dsICE)
```

```python
# Altitude bands per 250 m
elev_bands = np.arange(2500, 6001, 250)

#Grille pas exactement les mêmes valeurs des fois à ,e-5 près 
elev = elev.assign_coords({
    "X": RF.X,
    "Y": RF.Y
})

# RF mean per band of 250 m
RF_monthly_bands = RF_monthly_mean.groupby_bins(elev, bins=elev_bands).sum() / ICE_tot_bands
```

```python
#Grille pas exactement les mêmes valeurs des fois à ,e-5 près 
elev = elev.assign_coords({
    "X": SF.X,
    "Y": SF.Y
})

# SF mean per band of 250 m
SF_monthly_bands = SF_monthly_mean.groupby_bins(elev, bins=elev_bands).sum() / ICE_tot_bands
```

```python
#Grille pas exactement les mêmes valeurs des fois à ,e-5 près 
elev = elev.assign_coords({
    "X": TF.X,
    "Y": TF.Y
})

# TF mean per band of 250 m
TF_monthly_bands = TF_monthly_mean.groupby_bins(elev, bins=elev_bands).sum() / ICE_tot_bands
```

```python
R_monthly_bands = SF_monthly_bands / TF_monthly_bands * 100
```

```python
months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
y = [interval.right for interval in RF_monthly_bands.SH_bins.values]

fig, axes = plt.subplots(3, 4,figsize=(10, 16))
axes = axes.flatten()

for i, ax in enumerate(axes):
    ax.plot(
        RF_monthly_bands.sel(month=i+1).values, y, 
        marker='o', 
        linestyle='-',
        color='royalblue',
        label=f'RF',
        linewidth=2)
    
    ax.plot(
        SF_monthly_bands.sel(month=i+1).values, y, 
        marker='o', 
        linestyle='-',
        color='violet',
        label=f'SF',
        linewidth=2)
    
    ax.plot(
        TF_monthly_bands.sel(month=i+1).values, y, 
        marker='o', 
        linestyle='--',
        color='green',
        label=f'TF',
        linewidth=2, alpha = 0.6)
    
    ax.set_title(months[i], fontsize=16, weight="bold")
    
    if i in [0, 4, 8]:  # colonne gauche
        ax.set_ylabel('Elevation (m)', fontsize=12)
    else:
        ax.tick_params(labelleft=False)
        
    if i in [8, 9, 10, 11]:  # ligne du bas
        ax.set_xlabel('Precipitation (mm / month)', fontsize=12)
    
    ax.grid(True, linestyle='--', alpha=0.5, zorder=0)

handles, labels = axes[0].get_legend_handles_labels()
fig.legend(
    handles,
    labels,
    loc='lower center',
    bbox_to_anchor=(0.5, -0.04),
    ncol=3,
    fontsize=16,
    frameon=True
)    

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(3, 4,figsize=(10, 16))
axes = axes.flatten()

for i, ax in enumerate(axes):
    ax.plot(
        R_monthly_bands.sel(month=i+1).values, y, 
        marker='o', 
        linestyle='-',
        color='red',
        label=f'R',
        linewidth=2)
    
    ax.set_title(months[i], fontsize=16, weight="bold")
    
    if i in [0, 4, 8]:  # colonne gauche
        ax.set_ylabel('Elevation (m)', fontsize=12)
    else:
        ax.tick_params(labelleft=False)
        
    if i in [8, 9, 10, 11]:  # ligne du bas
        ax.set_xlabel('R (%)', fontsize=12)
    
    ax.grid(True, linestyle='--', alpha=0.5, zorder=0)

handles, labels = axes[0].get_legend_handles_labels()
fig.legend(
    handles,
    labels,
    loc='lower center',
    bbox_to_anchor=(0.5, -0.04),
    ncol=3,
    fontsize=16,
    frameon=True
)    

plt.tight_layout()
plt.show()
```
